```python
def move(inn,out,m):
    if not out or inn[-1] < out[-1]:
        out.append(inn.pop())
        return m+1
    return m
```

```python
def hanoi(n, inn, aux, out, m):
    if n == 0:
        return m

    m = hanoi(n-1, inn, out, aux, m)
    m = move(inn, out, m)
    m = hanoi(n-1, aux, inn, out, m)

    return m
```

```python
size = 20
tower1 = list(range(size,0,-1))
tower2, tower3 = [], []
draw()
moves = hanoi(size,tower1,tower2,tower3,0)
```
